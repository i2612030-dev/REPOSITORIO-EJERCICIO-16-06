```python
minutos = int(input("ingrese la duracion de la llamada en minutos: "))
domingo = input("es domingo? (si/no): ").lower()
turno = input(" turno (diurno/nocturno): ").lower()

costo = 0


if minutos <= 5:
  costo = minutos * 1.00

elif minutos <= 8:
  costo = 5 * 1.00 + (minutos - 5) * 0.8

elif minutos <= 10:
  costo = 5 * 1.00 + 3 * 0.8 + (minutos - 8) * 0.7

else:
  costo = 5 * 1.00 + 3 * 0.8 + 2 * 0.7 + (minutos - 10) * 0.5

if domingo == "si":
  impuesto = costo * 0.03
elif turno == "nocturno":
  impuesto = costo * 0.05
else:
  impuesto = 0

total = costo + impuesto
print("costo de la llamada: s/",round (costo, 2) )
print("impuesto: s/",round(impuesto, 2) ) # Corrected typo 'impuest' to 'impuesto'
print("total a pagar: s/",round(total, 2))
```

```python
cantidad = int(input("ingrese la cantidad de camisas: "))
precio = float(input("ingrese el precio de cada camisa: "))

subtotal = cantidad * precio

if cantidad >=1 and cantidad <= 4:
  descuento = subtotal * 0.125

elif cantidad >= 5 and cantidad <= 8:
  descuento = subtotal * 0.20

else:
  descuento = subtotal * 0.315

total = subtotal - descuento

print("compra sin descuento: s/", subtotal)
print("descuento: s/", descuento)
print("compra con descuento: s/", total)
```
